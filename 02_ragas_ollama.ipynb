```python

from datasets import Dataset
from ragas import evaluate
from ragas.metrics import (
    faithfulness,
    answer_relevancy,
    context_precision,
    context_recall,
)
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper
from langchain_ollama import ChatOllama, OllamaEmbeddings

# -------------------------------------------------
# 1. Configure local Ollama models
# -------------------------------------------------

ollama_llm = ChatOllama(
    model="llama3.2:latest",
    temperature=0,
)
ollama_llm = ChatOllama(
    model="gpt-oss:120b-cloud",
    temperature=0,
)

ollama_embeddings = OllamaEmbeddings(
    model="nomic-embed-text",
)

# Wrap LangChain models for RAGAS
evaluator_llm = LangchainLLMWrapper(ollama_llm)

evaluator_embeddings = LangchainEmbeddingsWrapper(
    ollama_embeddings
)

# -------------------------------------------------
# 2. Prepare RAG evaluation dataset
# -------------------------------------------------
# Each record must contain:
# question, answer, contexts, ground_truth

data = {
    "question": [
        "What is RAG in generative AI?",
        "What is the purpose of an embedding model?",
    ],
    "answer": [
        "RAG stands for Retrieval-Augmented Generation. It retrieves relevant information and provides it to an LLM to generate a grounded answer.",
        "An embedding model converts text into numerical vectors that represent semantic meaning.",
    ],
    "contexts": [
        [ "Retrieval-Augmented Generation (RAG) retrieves relevant documents from an external knowledge source and supplies them to a language model to help generate an answer." ],
        [ "Embedding models transform text into vectors that represent semantic information. These vectors are used for semantic search and similarity comparisons." ],
    ],
    "ground_truth": [
        "RAG retrieves relevant external information and uses it to help an LLM generate a grounded answer.",
        "An embedding model transforms text into numerical vectors representing semantic meaning.",
    ],
}

dataset = Dataset.from_dict(data)

# -------------------------------------------------
# 3. Select evaluation metrics
# -------------------------------------------------

metrics = [
    faithfulness,
    answer_relevancy,
    context_precision,
    context_recall,
]

# -------------------------------------------------
# 4. Evaluate using ONLY local Ollama models
# -------------------------------------------------

results = evaluate(
    dataset=dataset,
    metrics=metrics,
    llm=evaluator_llm,
    embeddings=evaluator_embeddings,
)

# -------------------------------------------------
# 5. Display and save results
# -------------------------------------------------

results_df = results.to_pandas()

print("\nRAGAS Evaluation Results:")
print(results_df.to_string(index=False))

print("\nAverage scores:")
print(results_df[
    [
        "faithfulness",
        "answer_relevancy",
        "context_precision",
        "context_recall",
    ]
].mean())

results_df.to_csv("ragas_results.csv", index=False)
print("\nResults saved to ragas_results.csv")
```
